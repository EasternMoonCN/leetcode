```python
from collections import deque
from typing import List
```

```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        # 数组
        queue = [] # 记录下表，需要判断队首是第几个元素，从而判断滑动窗口大小
        result = []
        for i, num in enumerate(nums):
            # 1. 入
            while queue and nums[queue[-1]] < num:
                queue.pop()
            queue.append(i)
            # 2. 出
            if i - queue[0] + 1 > k:
                queue.pop(0)
            # 3. 记录答案
            if i >= k - 1:
                result.append(nums[queue[0]])
        return result
```

```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        # 链表
        queue = deque() # 记录下表，需要判断队首是第几个元素，从而判断滑动窗口大小
        result = []
        for i, num in enumerate(nums):
            # 1. 入
            while queue and nums[queue[-1]] < num:
                queue.pop()
            queue.append(i)
            # 2. 出，每次只添加一个元素，所以只要判读一次有没有超过
            if i - queue[0] + 1 > k:
                queue.popleft()
            # 3. 记录答案
            if i >= k - 1:
                result.append(nums[queue[0]])
        return result
```
