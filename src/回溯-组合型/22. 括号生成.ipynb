```python
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        ans = []
        path = []
        # O(n*C(2n, n))
        def dfs(i: int, open: int) -> None:
            if i == 2*n:
                ans.append("".join(path))
                return
            # 选左
            if open < n:
                path.append('(')
                dfs(i + 1, open + 1)
                path.pop()
            # 不选左（右括号）
            close = i - open
            if close < open:
                path.append(')')
                dfs(i + 1, open)
                path.pop()
        dfs(0, 0)
        return ans
```

```python
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        ans = []
        path = [')'] * 2 * n
        def check():
            open = 0
            for c in path:
                if c == '(':
                    open += 1
                else:
                    open -= 1
                if open < 0:
                    return False
            return open == 0
        # i：下一个可选位置索引
        def dfs(i: int, open: int) -> None:
            # 不能用i==2n，这样必须要求2n-1这个位置是左括号
            # 另一种是我觉得不一定要当i==2n时，才会出现满足情况，比如012是左，就已经满足了
            # 而使用open==n，如果再继续+左，一定不对满足
            if open == n:
                if check():
                    ans.append("".join(path))
                return
            # 左括号选哪个位置
            for j in range(i, 2*n):
                path[j] = '('
                dfs(j + 1, open + 1)
                path[j] = ')'
        dfs(0, 0)
        return ans
```
