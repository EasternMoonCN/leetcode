给你一个字符串 s，请你将 s 分割成一些 子串，使每个子串都是 回文串 。返回 s 所有可能的分割方案。



示例 1：

输入：s = "aab"
输出：[["a","a","b"],["aa","b"]]
示例 2：

输入：s = "a"
输出：[["a"]]


提示：

1 <= s.length <= 16
s 仅由小写英文字母组成

```python
class Solution:
    def partition(self, s: str) -> List[List[str]]:
        ans = []
        path = []
        n = len(s)
        # 子串[i, j+1] 选不选
        def dfs(i, j):
            if j == n:
                if i == j:  # 说明最后一次也满足,不满足情况下，说明最后一次是从不选过来的
                    ans.append(path.copy())
                return
            # 选
            c = s[i: j + 1]
            if c == c[::-1]:    # 回文
                path.append(c)
                dfs(j + 1, j + 1)
                path.pop()
            # 不选
            dfs(i, j + 1)
        dfs(0, 0)
        return ans
```

```python
from typing import List

class Solution:
    def partition(self, s: str) -> List[List[str]]:
        ans = []
        n = len(s)
        def dfs(i: int, j: int, path: List[str]) -> None:
            # 所有字符串划分后，才是一种解
            if i == n:
                ans.append(path.copy())
                return
            if j > n:
                return
            # 选
            c = s[i: j + 1]
            if c == c[::-1]:
                dfs(j + 1, j + 1, path + [c])
            # 不选
            dfs(i, j + 1, path)
        dfs(0, 0, [])
        return ans
if __name__ == '__main__':
    print(Solution().partition("aab"))
```
