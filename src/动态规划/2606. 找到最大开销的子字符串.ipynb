```python
class Solution:
    def maximumCostSubstring(self, s: str, chars: str, vals: List[int]) -> int:
        d = dict()
        for k, v in zip(chars, vals):
            d[k] = v
        n = len(s)
        # 以i结尾最大开销
        dp = [-inf] + [0] * n
        max_cost = 0 # 空串
        for i, c in enumerate(s):
            cost = d[c] if c in d else (ord(c) - ord('a') + 1)
            dp[i + 1] = max(cost, dp[i] + cost)
            max_cost = max(max_cost, dp[i + 1])
        return max_cost
```

```python
class Solution:
    def maximumCostSubstring(self, s: str, chars: str, vals: List[int]) -> int:
        d = dict()
        for k, v in zip(chars, vals):
            d[k] = v
        pre_cost = 0
        max_cost = 0 # 空串
        min_pre_cost = 0
        for i, c in enumerate(s):
            cost = d[c] if c in d else (ord(c) - ord('a') + 1)
            # 前缀和
            pre_cost += cost
            max_cost = max(max_cost, pre_cost - min_pre_cost)
            # 最小前缀
            min_pre_cost = min(min_pre_cost, pre_cost)
        return max_cost
```
