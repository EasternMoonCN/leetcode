```python
from typing import List
from math import inf
```

```python
class Solution:
    # 从target中去掉k个nums数组后，剩下余数从nums + nums找rem
    def minSizeSubarray(self, nums: List[int], target: int) -> int:
        n = len(nums)
        total = sum(nums)
        rem = target % total
        left = 0
        res = inf
        for right in range(n * 2):
            rem -= nums[right % n]
            while rem < 0:
                rem += nums[left % n]
                left += 1
            if rem == 0:
                res = min(res, right - left + 1)
        # target // total个nums数组，每个个数为n
        return res + target // total * n if res < inf else -1
```
