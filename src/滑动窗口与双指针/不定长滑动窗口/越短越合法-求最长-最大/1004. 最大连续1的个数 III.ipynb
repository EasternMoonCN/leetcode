```python
from typing import List
```

```python
class Solution:
    def longestOnes(self, nums: List[int], k: int) -> int:
        ans = left = res = 0
        for right, num in enumerate(nums):
            if num == 0:
                res += 1
            while res > k:
                if nums[left] == 0:
                    res -= 1
                left += 1
            ans = max(ans, right - left + 1)
        return ans
```
