```python
from typing import List
```

```python
class Solution:
    # 不能用k去除，因为k先除会遇到小数
    def numSubarrayProductLessThanK(self, nums: List[int], k: int) -> int:
        if k <= 1:
            return 0
        left = res = 0
        prod = 1
        for right, num in enumerate(nums):
            prod *= num
            while prod >= k:
                prod /= nums[left]
                left += 1
            # [left, right] 满足要求时，有[left right]， [left + 1 right]，..., [right - 1 right]
            # 为什么不要考虑[left, left + 1] [left, left + 2]?
            # 因为 [left, left + 1] 在right右端点为left+1时，已经计算过了
            res += right - left + 1
        return res
```

```python
class Solution:
    # 不能用k去除，因为k先除会遇到小数
    def numSubarrayProductLessThanK(self, nums: List[int], k: int) -> int:
        left = res = 0
        prod = 1
        for right, num in enumerate(nums):
            prod *= num
            while left <= right and prod >= k:
                prod /= nums[left]
                left += 1
            # [left, right] 满足要求时，有[left right]， [left + 1 right]，..., [right - 1 right]
            # 为什么不要考虑[left, left + 1] [left, left + 2]?
            # 因为 [left, left + 1] 在right右端点为left+1时，已经计算过了
            res += right - left + 1
        return res
```
