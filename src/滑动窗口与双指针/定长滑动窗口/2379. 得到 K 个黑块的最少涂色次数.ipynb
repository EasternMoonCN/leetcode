```python
class Solution:
    def minimumRecolors(self, blocks: str, k: int) -> int:
        res = 0
        ans = len(blocks)
        for i, block in enumerate(blocks):
            if block == 'W':
                res += 1
            # 长度不够k
            if i + 1 < k:
                continue
            ans = min(ans, res)
            # k个，队头为 i-k+1
            if blocks[i - k + 1] == 'W':
                res -= 1
        return ans
```
