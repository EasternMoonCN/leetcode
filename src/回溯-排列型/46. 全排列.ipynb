```python
from typing import List, Set
class Solution:
    # O(n*n!)：叶子结点个数
    def permute(self, nums: List[int]) -> List[List[int]]:
        n = len(nums)
        ans = []
        def dfs(path: Set[int]) -> None:
            if len(path) == n:
                ans.append(path.copy())
                return
            for num in nums:
                if num in path:
                    continue
                path.add(num)
                dfs(path)
                path.pop()
        dfs(set())
        return ans
```

```python
class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:
        n = len(nums)
        ans = []
        path = []
        flag = [False] * n
        def dfs() -> None:
            if len(path) == n:
                ans.append(path.copy())
                return
            for i, num in enumerate(nums):
                if flag[i]:
                    continue
                flag[i] = True
                path.append(num)
                dfs()
                flag[i] = False
                path.pop()
        dfs()
        return ans
```
