```python
class Solution:
    def wordSquares(self, words: List[str]) -> List[List[str]]:
        n = len(words)
        words.sort()
        is_choice = [False] * n
        ans = []
        path = []
        def dfs(i: int) -> None:
            if len(path) == 4:
                top, left, right, bottom = path
                if top[0] == left[0] and top[3] == right[0] and bottom[0] == left[3] and bottom[3] == right[3]:
                    ans.append(path + [])
                return
            if i == n:
                return
            # 选
            if not is_choice[i]:
                is_choice[i] = True
                path.append(words[i])
                dfs(0)
                path.pop()
                is_choice[i] = False

            dfs(i + 1)
        dfs(0)
        return ans
```
