```python
from collections import defaultdict
```

```python
class Solution:
    def maxConsecutiveAnswers(self, answerKey: str, k: int) -> int:
        def maxRes(answerKey: str, k: int, c: chr) -> int:
            # 窗口内最多有两个c
            ans = left = res = 0
            for right, s in enumerate(answerKey):
                if s == c:
                    res += 1
                while res > k:
                    out = answerKey[left]
                    if out == c:
                        res -= 1
                    left += 1
                ans = max(ans, right - left + 1)
            return ans

        return max(maxRes(answerKey, k, 'T'), maxRes(answerKey, k, 'F'))

```

```python
class Solution:
    def maxConsecutiveAnswers(self, answerKey: str, k: int) -> int:
        ans = left = 0
        idict = defaultdict(int)
        for right, ch in enumerate(answerKey):
            idict[ch] += 1
            # 窗口内两个都不满足时，必须要出窗口
            # 只有一个（如T）不满足的情况下，说明宁一个（F）满足，可以将F转换为T
            while idict['T'] > k and idict['F'] > k:
                idict[answerKey[left]] -= 1
                left += 1
            ans = max(ans, right - left + 1)
        return ans


```
