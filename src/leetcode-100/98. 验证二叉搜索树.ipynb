```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        res = []
        def dfs(root):
            if root is None:
                return
            dfs(root.left)
            res.append(root.val)
            dfs(root.right)
        dfs(root)
        for i in range(1, len(res)):
            if res[i] <= res[i - 1]:
                return False
        return True
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    # 左子树的最后一个节点
    pre = -inf
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        if root is None:
            return True
        # 左子树
        if not self.isValidBST(root.left):
            return False
        # 节点值
        if root.val <= self.pre:
            return False
        # 这里记录的其实是当前root节点的前一个访问的节点值
        self.pre = root.val
        # 右子树
        return self.isValidBST(root.right)
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    # 左子树的最后一个节点
    pre = -inf
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        def dfs(root):
            if root is None:
                # 所有节点都满足
                return inf, -inf
            l_min, l_max = dfs(root.left)
            r_min, r_max = dfs(root.right)
            # inf, -inf 可以保证该条件一定不成立
            # -inf， inf 保证该条件一定成立
            x = root.val
            if x <= l_max or x >= r_min:
                return -inf, inf
            return min(l_min, x), max(x, r_max)
        return dfs(root)[1] != inf
```
