```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        cque = deque()
        ans = 0
        for c in s:
            # 将所有相同元素出队列
            while c in cque:
                cque.popleft()
            cque.append(c)
            ans = max(ans, len(cque))
        return ans
```

```python
class Solution:
    # O(N)
    def lengthOfLongestSubstring(self, s: str) -> int:
        cque = list()
        ans = 0
        for c in s:
            # 将所有相同元素出队列
            while c in cque:
                cque.popleft()
            cque.append(c)
            ans = max(ans, len(cque))
        return ans
```

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        cset = set()
        ans = 0
        left = 0
        for c in s:
            # 将所有相同元素出队列
            while c in cset:
                cset.remove(s[left])
                left += 1
            cset.add(c)
            ans = max(ans, len(cset))
        return ans
```
