```python
class Solution:
    def maximumUniqueSubarray(self, nums: List[int]) -> int:
        iset = set()
        res = ans = left = 0
        for right, num in enumerate(nums):
            while num in iset:
                iset.remove(nums[left])
                res -= nums[left]
                left += 1
            iset.add(num)
            res += num
            ans = max(ans, res)
        return ans
```
