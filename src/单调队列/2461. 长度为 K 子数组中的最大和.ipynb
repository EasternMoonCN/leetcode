```python
from typing import List
from collections import defaultdict
```

```python
class Solution:
    def maximumSubarraySum(self, nums: List[int], k: int) -> int:
        idict = defaultdict(int)
        ans = res = 0
        for i, num in enumerate(nums):
            idict[num] += 1
            res += num
            if i + 1 < k:
                continue
            if len(idict) == k:
                ans = max(ans, res)
            out = nums[i - k + 1]
            res -= out
            idict[out] -= 1
            if idict[out] == 0:
                # idict.pop(out)
                del idict[out]
        return ans
```
