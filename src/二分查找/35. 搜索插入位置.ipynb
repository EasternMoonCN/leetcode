```python
from typing import List
```

```python
class Solution:
    def lower_bound(self, nums: List[int], target: int) -> int:
        # [l, h]
        l, h = 0, len(nums) - 1
        while l <= h:
            m = (l + h) // 2
            if nums[m] < target:
                l = m + 1
            else:   # nums[m] >= target
                h = m - 1
        return l


    def searchInsert(self, nums: List[int], target: int) -> int:
        return self.lower_bound(nums, target)
```
