```python
idict = defaultdict(int)
print(idict[0])
ldict = defaultdict(list)
print(ldict[0])
ddict = defaultdict(dict)
print(ddict[0])
```

```python
class Solution:
    def maxSum(self, nums: List[int], m: int, k: int) -> int:
        ans = 0
        idict = {}
        res = 0
        for i, num in enumerate(nums):
            idict[num] = idict[num] + 1 if num in idict else 1
            res += num
            if i + 1 < k:
                continue
            if len(idict) >= m:
                ans = max(ans, res)
            # [i - k + 1, i] 长度为k
            out = nums[i - k + 1]
            res -= out
            idict[out] = idict[out] - 1
            if idict[out] == 0:
                idict.pop(out)
        return ans


```

```python
if __name__ == '__main__':
    nums = [2, 6, 7, 3, 1, 7]
    m = 3
    k = 4
    print(Solution().maxSum(nums, m, k))
```
