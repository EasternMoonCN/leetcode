```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        # 不需要考虑root为None，首先根节点非none，其次pq存在左右子树，存在情况下才便利
        if root.val > p.val and root.val > q.val:   # pq都在左子树中
            return self.lowestCommonAncestor(root.left, p, q)
        if root.val < p.val and root.val < q.val:   # pq都在右子树中
            return self.lowestCommonAncestor(root.right, p, q)
        return root # pq在两侧
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    # 倒序，左右都有，或左右存在一个，且当前是 就 更新
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        ans = [None]
        # 判断root是否p或q
        def dfs(root):
            if root is None:
                return False
            if root.val > p.val and root.val > q.val:
                left = dfs(root.left)
                right = False
            elif root.val < p.val and root.val < q.val:
                right = dfs(root.right)
                left = False
            else:
                left = dfs(root.left)
                right = dfs(root.right)
            if (left and right) or ((left or right) and (root is p or root is q)):
                ans[0] = root
            return left or right or root is p or root is q
        dfs(root)
        return ans[0]
```
