```python
class Solution:
    def equalSubstring(self, s: str, t: str, maxCost: int) -> int:
        ideque = deque()
        ans = 0
        for cs, ct in zip(s, t):
            cost = abs(ord(cs) - ord(ct))
            ideque.append(cost)
            maxCost -= cost
            while maxCost < 0:
                cost = ideque.popleft()
                maxCost += cost
            ans = max(ans, len(ideque))
        return ans
```

```python
class Solution:
    def equalSubstring(self, s: str, t: str, maxCost: int) -> int:
        ans = left = 0
        for right, (cs, ct) in enumerate(zip(s, t)):
            cost = abs(ord(cs) - ord(ct))
            maxCost -= cost
            while maxCost < 0:
                cost = abs(ord(s[left]) - ord(t[left]))
                left += 1
                maxCost += cost
            ans = max(ans, right - left + 1)
        return ans
```
