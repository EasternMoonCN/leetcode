```python
from typing import List
from math import inf
```

动态规划思想

```python
class Solution:
    def maxResult(self, nums: List[int], k: int) -> int:
        n = len(nums)
        dp = [0] * n
        for i in range(n):
            tmp = -inf
            for j in range(1, k + 1):
                if i - j >= 0:
                    # 是正的一定会经过
                    if nums[i - j] >= 0:
                        tmp = dp[i - j]
                        break
                    tmp = max(tmp, dp[i - j])
            dp[i] = nums[i] if tmp == -inf else tmp + nums[i]
        return dp[n - 1]
```

```python
from collections import deque
```

```python
class Solution:
    def maxResult(self, nums: List[int], k: int) -> int:
        # 在大小为k的窗口递增
        q = deque([0])
        n = len(nums)
        dp = [0] * n
        # 和
        dp[0] = nums[0]

        for i in range(1, n):
            # 超过窗口大小
            if q[0] + k  < i:
                q.popleft()
            # 队头元素是窗口内和最大的
            if q:
                dp[i] = nums[i] + dp[q[0]]
            # 当前i肯定是找窗口内和最大的
            # 反之不行，会把窗口内的较大的和值移除
            while q and dp[q[-1]] <= dp[i]:
                q.pop()
            q.append(i)
        return dp[-1]
```
