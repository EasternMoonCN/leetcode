```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        # dp[i][j]：nums i 到j的最大数组和
        # -2 1 -3
        #    1
        n = len(nums)
        maxV = -10**4
        maxCols = [-10**4] * n
        dp = [[0] * n for _ in range(n)]
        for i in range(n):
            for j in range(i, n):
                if j == 0:
                    maxJ = 0
                else:
                    maxJ = maxCols[j - 1]
                # maxJ = max([row[j - 1] for row in dp])
                # maxJ = maxCols[j - 1]
                dp[i][j] = maxJ + nums[j] if maxJ > 0 else nums[j]
                maxV = dp[i][j] if dp[i][j] > maxV else maxV
                if dp[i][j] > maxCols[j]:
                    maxCols[j] = dp[i][j]
        return maxV
```

```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        n = len(nums)
        res = -inf
        # 前缀和
        pre = 0
        # 最小前缀和
        min_pre = 0
        for x in nums:
            pre += x
            res = max(res, pre - min_pre)
            # 先不选
            min_pre = min(min_pre, pre)
        return res
```
