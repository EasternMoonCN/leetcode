```python
class Solution:
    # 第i个位置节水量 = min(左前缀最大值， 右前缀最大值) - 当前值
    def trap(self, height: List[int]) -> int:
        n = len(height)
        i, j = 0, n - 1
        cur_height, ans = 0, 0
        while i <= j:
            if cur_height - min(height[i], height[j]) < 0:
                ans += (min(height[i], height[j]) - cur_height) * (j - i - 1)
            if height[i] < height[j]:
                if i != 0:
                    ans -= min(cur_height, height[i])
                cur_height = max(cur_height, height[i])
                i += 1
            else:
                if j != n - 1:
                    ans -= min(cur_height, height[j])
                cur_height = max(cur_height, height[j])
                j -= 1
        return ans
```

```python
class Solution:
    # 第i个位置节水量 = min(左前缀最大值， 右前缀最大值) - 当前值
    def trap(self, height: List[int]) -> int:
        n = len(height)
        i, j = 0, n - 1
        pre_max, suf_max, ans = 0, 0, 0
        while i < j:
            pre_max = max(pre_max, height[i])
            suf_max = max(suf_max, height[j])
            if pre_max < suf_max:
                ans += pre_max - height[i]
                i += 1
            else:
                ans += suf_max - height[j]
                j -= 1
        return ans
```
