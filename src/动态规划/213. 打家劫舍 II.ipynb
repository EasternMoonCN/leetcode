```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        if n == 1:
            return nums[0]
        dp = [[0] * 2 for _ in range (n + 2)]
        for i in range(n):
            for j in range(2):
                if i == 0 and j == 0:
                    continue
                dp[i + 2][j] = max(dp[i + 1][j], dp[i][j] + nums[i])
        return max(dp[n][1], dp[n + 1][0])
```
