```python
class Solution:
    def maximumLengthSubstring(self, s: str) -> int:
        ans = 0
        # cdict = dict()
        cdict = {}
        # [left right]之间可以看作一个队列，其实是一个窗口
        left = 0
        for right, c in enumerate(s):
            while c in cdict and cdict[c] == 2:
                cdict[s[left]] -= 1
                left += 1
            cdict[c] = cdict[c] + 1 if c in cdict else 1
            ans = max(ans, right - left + 1)
        return ans
```
