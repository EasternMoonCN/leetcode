```python
from typing import List
from math import inf
```

```python
class Solution:
    def findMaxValueOfEquation(self, points: List[List[int]], k: int) -> int:
        n = len(points)
        ans = -inf
        for i in range(n):
            for j in range(i + 1, n):
                xi = points[i][0]
                xj = points[j][0]
                if abs(xi - xj) <= k:
                    yi = points[i][1]
                    yj = points[j][1]
                    ans = max(ans, yi + yj + abs(xi - xj))
        return ans
```

```python
from collections import deque
```

- yi + yj + |xi - xj|
- = yi + yj + xj - xi
- = (xj + yj) + (yi - xi)

```python
class Solution:
    def findMaxValueOfEquation(self, points: List[List[int]], k: int) -> int:
        q = deque()
        ans = -inf
        for x, y in points:
            # 队头超过范围了
            while q and x - q[0][0] > k:
                q.popleft()
            # 需要先计算答案，不能在比较完队尾后计算
            if q:
                # if ans < point[0] + point[1] + q[-1][1]:
                #     print(f"{point}, {q[-1]}, {point[0] + point[1] + q[-1][1]}")
                ans = max(ans, x + y + q[0][1])
            # 队尾元素没有新来的(y - x)大
            while q and q[-1][1] <= y - x:
                q.pop()
            q.append([x, y - x])
        return ans
```
