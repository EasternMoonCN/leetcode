```python
class Solution:
    ans = 0
    def totalNQueens(self, n: int) -> int:
        cols = [False] * n
        fdiag = [False] * (2 * n - 1) # j + i [0, 2n - 2]
        sdiag = [False] * (2 * n - 1) # j - i [-n + 1, n - 1]
        def dfs(i: int):
            if i == n:
                self.ans += 1
            for j in range(n):
                if cols[j] or fdiag[j + i] or sdiag[j - i]:
                    continue
                cols[j] = fdiag[j + i] = sdiag[j - i] = True
                dfs(i + 1)
                cols[j] = fdiag[j + i] = sdiag[j - i] = False
        dfs(0)
        return self.ans
```
