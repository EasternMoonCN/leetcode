```python
from typing import List
```

```python
class Solution:
    # >=
    def lower_bound(self, nums: List[int], target: int) -> int:
        # [low, height]
        low, height = 0, len(nums) - 1
        # mid = (low + height) >> 1
        # 写等于是为了判断 target 位于边界情况
        while low <= height:
            # 向下取证
            mid = (low + height) >> 1
            # mid = (low + height) // 2
            if nums[mid] < target:
                low = mid + 1
            else:
                height = mid - 1
        # 在height满足 >= target的情况下，还在执行循环，直到不满足，所以取low
        return low

    def searchRange(self, nums: List[int], target: int) -> List[int]:
        left = self.lower_bound(nums, target)
        if left == len(nums):
            return [-1, -1]
        # <= x  ==>   (>= x+1) -1
        right = self.lower_bound(nums, target + 1) - 1
        if left > right:
            return [-1, -1]
        return [left, right]
```

```python
if __name__ == '__main__':
    so = Solution()
    print(so.searchRange([5, 7, 7, 8, 9], 7))
```
