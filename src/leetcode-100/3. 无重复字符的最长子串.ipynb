```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        cset = set()
        ans = j = 0

        for i in range(len(s)):
            c = s[i]
            while c in cset:
                cset.remove(s[j])
                j += 1
            cset.add(c)
            ans = max(ans, len(cset))
        return ans
```

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        cset = set()
        ans = i = j = 0

        while i < len(s):
            c = s[i]
            i += 1
            while c in cset:
                cset.remove(s[j])
                j += 1
            cset.add(c)
            ans = max(ans, i - j)
        return ans
```
