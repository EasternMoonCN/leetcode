定长

```python
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        pdict = defaultdict(int)
        for c in p:
            pdict[c] += 1

        sdict = defaultdict(int)
        n = len(p)
        left = 0
        ans = []
        for right, c in enumerate(s):
            ## 尾进
            sdict[c] += 1
            ## 看看长度满不满足
            if right - left + 1 < n:
                continue
            ## 长度满足情况下，看看元素一样吗
            if sdict == pdict:
                ans.append(left)
            ## 头出
            sdict[s[left]] -= 1
            if sdict[s[left]] == 0:
                sdict.pop(s[left])
            left += 1
        return ans
```

```python
class Solution:
    #O(n + m)
    def findAnagrams(self, s: str, p: str) -> List[int]:
        pcnt = Counter(p)
        scnt = Counter()

        n = len(p)
        left = 0
        ans = []
        for right, c in enumerate(s):
            ## 尾进
            scnt[c] += 1
            ## 看看长度满不满足
            if right - left + 1 < n:
                continue
            ## 长度满足情况下，看看元素一样吗
            if scnt == pcnt:
                ans.append(left)
            ## 头出
            scnt[s[left]] -= 1
            left += 1
        return ans
```

不定长

```python
class Solution:
    #
    def findAnagrams(self, s: str, p: str) -> List[int]:
        pcnt = Counter(p)
        left = 0
        ans = []
        for right, c in enumerate(s):
            ## c先从right进入窗口
            pcnt[c] -= 1
            ## c太多了，left右边出窗口
            while pcnt[c] < 0:
                pcnt[s[left]] += 1
                left += 1
            ## 字符c个数满足，且长度满足，则一定想等
            if right - left + 1 == len(p):
                ans.append(left)
        return ans
```
