```python
class Solution:
    def splitString(self, s: str) -> bool:
        n = len(s)
        def dfs(i: int, path: int) -> bool:
            if i == n:
                # 确保分割过
                return path != int(s)
            for j in range(i, n):
                c = int(s[i: j + 1])
                if path == -1 or path - 1 == c:
                    if dfs(j + 1, c):
                        return True
            return False
        return dfs(0, -1)
```

```python
class Solution:
    def splitString(self, s: str) -> bool:
        n = len(s)
        def dfs(i: int, pre: int, cnt: int) -> bool:
            if i == n:
                return cnt > 1
            cur = 0
            for j in range(i, n):
                cur = cur * 10 + int(s[j])
                if pre == -1 or pre - 1 == cur:
                    if dfs(j + 1, cur, cnt + 1):
                        return True
            return False
        return dfs(0, -1, 0)
```
