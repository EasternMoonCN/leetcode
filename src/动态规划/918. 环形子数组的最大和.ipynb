```python
class Solution:
    def maxSubarraySumCircular(self, nums: List[int]) -> int:
        n = len(nums)
        dp_max = [0] * (n + 1)
        dp_min = [0] * (n + 1)
        res_max = -inf  # 不允许最大子数组为空
        res_min = 0  # 允许最小子数组为空
        for i in range(n):
            # 最大子数组和
            dp_max[i + 1] = max(dp_max[i] + nums[i], nums[i])
            res_max = max(res_max, dp_max[i + 1])

            # 最小子数组和（最大数组和在旁边，则中间的最小）
            dp_min[i + 1] = min(dp_min[i] + nums[i], nums[i])
            res_min = min(res_min, dp_min[i + 1])
        # 数组全部为负时，就过就是最大的负数
        if res_max < 0:
            return res_max
        # 要么在中间，要么是所有的减去中间最小的
        return max(res_max, sum(nums) - res_min)
```
