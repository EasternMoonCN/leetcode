```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        cur = headA
        a_set = set()

        while cur:
            a_set.add(cur)
            cur = cur.next

        cur = headB
        while cur:
            if cur in a_set:
                return cur
            cur = cur.next
        return None
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    # A : x + z
    # B : y + z
    # 如果x=y，则直接在z1相遇
    # 如果x > y，且有相遇点, 则 x + z + y = y + z + x，在z1相遇（反之一样）
    # 如果无相遇点，x + y = y  + x，在空点相遇
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        p, q = headA, headB
        while p != q:
            p = p.next
            q = q.next
            if p is None and q is None:
                return None
            if p is None:
                p = headB
            if q is None:
                q = headA
        return p
```
