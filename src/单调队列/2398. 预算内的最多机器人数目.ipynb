```python
from typing import List
from collections import deque
```

chargeTimes = [3,6,1,3,4], runningCosts = [2,1,3,4,5], budget = 25

- 维护了一个左右指针
- left指向左边元素，right指向当前遍历元素
- 队列q维护了窗口内[left, right]的单调递减队列
- s维护了[left, right]之间runningCosts和

```python
class Solution:
    def maximumRobots(self, chargeTimes: List[int], runningCosts: List[int], budget: int) -> int:
        q = deque()
        ans = s = left = 0
        for right, (c, r) in enumerate(zip(chargeTimes, runningCosts)):
            # 1. 入
            while q and chargeTimes[q[-1]] <= c:
                q.pop()
            q.append(right)
            s += r
            # 2. 出
            while q and chargeTimes[q[0]] + (right - left + 1) * s > budget:
                # 下表相等时才移除
                if q[0] == left:
                    q.popleft()
                s -= runningCosts[left]
                left += 1
            # 3. 答案
            if q:
                ans = max(right - left +  1, ans)
        return ans
```

```python
if __name__ == '__main__':
    print(Solution().maximumRobots(chargeTimes=[3, 6, 1, 3, 4], runningCosts=[2, 1, 3, 4, 5], budget=25))
```
