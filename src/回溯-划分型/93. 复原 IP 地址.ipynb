```python
from typing import List
```

```python
class Solution:
    def restoreIpAddresses(self, s: str) -> List[str]:
        ans = []
        n = len(s)
        def dfs(i: int, path: str) -> None:
            if i == n:
                if path.count(".") == 3:
                    ans.append(path)
                return
            for j in range(i, n):
                if path.count(".") > 2:
                    return
                c = s[i: j + 1]
                if c == '0' or (c[0] != '0' and int(c) >= 0 and int(c) <= 255):
                    dfs(j + 1, path + '.' + c if path != "" else c)
        dfs(0, "")
        return ans
```

```python
class Solution:
    def restoreIpAddresses(self, s: str) -> List[str]:
        ans = []
        n = len(s)
        def dfs(i: int, path: List[str], cnt: int) -> None:
            # 每个至少1位 剩余字符数量 每个至多3位
            # 剪枝，并确保了合法长度（只会分割4个）
            if not 4 - cnt <=  n - i <= 3 * (4 - cnt):
                return
            if i == n:
                ans.append(".".join(path))
                return
            val = 0
            for j in range(i, n):
                val = val * 10 + int(s[j])
                if val > 255:
                    break
                dfs(j + 1, path + [s[i: j + 1]], cnt + 1)
                if val == 0:    # 前导0：0xx
                    break
        dfs(0, [], 0)
        return ans
```
