```python
class Solution:
    ## 无法保持非0元素相对位置,这里是移0
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        l = 0; r = len(nums) - 1
        while l < r:
            if nums[l] == 0:
                nums[l] = nums[r]
                nums[r] = 0
                r -= 1
            l += 1
```

```python
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        tmp = [0] * len(nums)
        i = 0
        for num in nums:
            if num != 0:
                tmp[i] = num
                i += 1
        nums.clear()
        nums.extend(tmp)
```

```python
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        ## 栈
        i = 0
        for j, num in enumerate(nums):
            if num != 0:
                # 入栈
                nums[i] = num
                i += 1
        nums[i:] = [0] * (len(nums) - i)
```
