```python
class NumArray:

    def __init__(self, nums: List[int]):
        n = len(nums)
        self.pre_sum = [0] * n
        for i in range(n):
            self.pre_sum[i] = self.pre_sum[i - 1] + nums[i]
        # print(self.pre_sum)

    def sumRange(self, left: int, right: int) -> int:
        # print(f"{self.pre_sum[right]}, {0 if left == 0 else self.pre_sum[left - 1]}")
        return self.pre_sum[right] - (0 if left == 0 else self.pre_sum[left - 1])



# Your NumArray object will be instantiated and called as such:
# obj = NumArray(nums)
# param_1 = obj.sumRange(left,right)
```

```python
class NumArray:

    def __init__(self, nums: List[int]):
        n = len(nums)
        ## pre_sum[i] ==> [0, i - 1]
        self.pre_sum = [0] * (n + 1)
        for i in range(n):
            self.pre_sum[i + 1] = self.pre_sum[i] + nums[i]

    def sumRange(self, left: int, right: int) -> int:
        ##self.pre_sum[left] ==> [0, left - 1]
        ## self.pre_sum[right + 1] ==> [0, right]
        return self.pre_sum[right + 1] - self.pre_sum[left]



# Your NumArray object will be instantiated and called as such:
# obj = NumArray(nums)
# param_1 = obj.sumRange(left,right)
```
