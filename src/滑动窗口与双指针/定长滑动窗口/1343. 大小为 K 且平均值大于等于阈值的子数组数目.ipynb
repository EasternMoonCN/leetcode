```python
class Solution:
    def numOfSubarrays(self, arr: List[int], k: int, threshold: int) -> int:
        left = 0
        res = 0
        ans = 0
        for right, num in enumerate(arr):
            # 长度超了
            if right - left + 1 > k:
                res -= arr[left]
                left += 1
            res += num
            # 有k个元素
            if right + 1 >= k and res / k >= threshold:
                ans += 1
        return ans
```

```python
class Solution:
    def numOfSubarrays(self, arr: List[int], k: int, threshold: int) -> int:
        res = 0
        ans = 0
        for right, num in enumerate(arr):
            # [right - k, right - 1]满了k个，需要留出来一个给num
            if right >= k:
                res -= arr[right - k]
            # 将当前位置right添加到集合
            res += num
            # [right - k + 1, right]要有k个
            if right + 1 >= k and res / k >= threshold:
                ans += 1
        return ans
```

```python
class Solution:
    def numOfSubarrays(self, arr: List[int], k: int, threshold: int) -> int:
        res = ans = 0
        for right, num in enumerate(arr):
            # 将right添加到集合，此时集合中[right - k + 1, right]
            res += num
            # 至少有k个元素，即right - k + 1 < 0 时没有k个元素
            if right + 1 < k:
                continue
            if res / k >= threshold:
                ans += 1
            # 空出right-k+1的位置为下一次添加元素
            res -= arr[right - k + 1]
        return ans
```
