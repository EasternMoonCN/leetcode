```python
from typing import List
from typing import Optional
```

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
class Solution:
    def nextLargerNodes(self, head: Optional[ListNode]) -> List[int]:
        ans = []
        stack = []
        while head:
            # 这些元素找到了下一个更大值
            while stack and stack[-1][0].val < head.val:
                ans[stack.pop()[1]] = head.val
            stack.append([head, len(ans)])
            ans.append(0) # 遍历到长度，用于占位
            head = head.next
        return ans
```

- 递归
    - 递：求链表长度，初始化ans数组
    - 归：反向计算ans

```python
class Solution:
    def nextLargerNodes(self, head: Optional[ListNode]) -> List[int]:
        ans = []
        stack = []

        def dfs(node: Optional[ListNode], i: int) -> None:
            if not node:
                return
            ans.append(0)
            dfs(node.next, i + 1)
            while stack and stack[-1] <= node.val:
                stack.pop()
            if stack:
                ans[i] = stack[-1]
            stack.append(node.val)

        dfs(head, 0)
        return ans
```

```python
if __name__ == '__main__':
    solution = Solution()
    print(solution.nextLargerNodes(ListNode(2, ListNode(1, ListNode(5)))))
    print(solution.nextLargerNodes(ListNode(2, ListNode(7, ListNode(4, ListNode(3, ListNode(5)))))))
```
