```python
from typing import List
from functools import cache
```

```python
class Solution:
    def combinationSum4(self, nums: List[int], target: int) -> int:
        n = len(nums)
        @cache
        def dfs(t: int) -> int:
            if t == 0:
                return 1
            cnt = 0
            for i in range(n):
                if t >= nums[i]:
                    cnt += dfs(t - nums[i])
            return cnt
        return dfs(target)
```
