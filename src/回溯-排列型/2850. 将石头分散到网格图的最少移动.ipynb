```python
from itertools import permutations
from typing import List
from math import inf
class Solution:
    # {i, j}多 到 {i, j}空 的最短哈夫曼距离
    def minimumMoves(self, grid: List[List[int]]) -> int:
        source = []
        to = []
        rows = len(grid)
        cols = len(grid[0])
        for i in range(rows):
            for j in range(cols):
                if grid[i][j] > 1:
                    for _ in range(1, grid[i][j]):
                        source.append([i, j])
                elif grid[i][j] == 0:
                    to.append([i, j])
        # 求source到to全排列的最短曼哈顿距离
        # print(f"{source}")
        # print(f"{to}")

        per = list(permutations(source))
        n = len(source)
        ans = inf
        for s in per:
            tmp = 0
            for i in range(n):
                tmp += abs(s[i][0] - to[i][0]) + abs(s[i][1] - to[i][1])
            ans = min(ans, tmp)
        return ans

```
