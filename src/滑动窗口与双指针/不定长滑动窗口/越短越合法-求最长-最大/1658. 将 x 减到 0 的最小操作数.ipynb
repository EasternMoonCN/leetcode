```python
from typing import List
```

```python
class Solution:
    def minOperations(self, nums: List[int], x: int) -> int:
        target = sum(nums) - x
        if target < 0:
            return -1
        if target == 0:
            return len(nums)
        # 求最长
        ans = res = left = 0
        for right, num in enumerate(nums):
            res += num
            while res > target:
                res -= nums[left]
                left += 1
            if res == target:
                ans = max(ans, right - left + 1)
        return -1 if ans == 0 else len(nums) - ans
```
