```python
class Solution:
    # [-4, -1, -1, 0, 1, 2]
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        n = len(nums)
        ans = list()
        for i in range(n):
            if i != 0 and nums[i] == nums[i - 1]:
                continue
            for j in range(i + 1, n):
                if j != i + 1 and nums[j] == nums[j - 1]:
                    continue
                t = -nums[i] - nums[j]
                k = self.binaray_search(nums, t, j + 1)
                if k != -1:
                    ans.append([nums[i], nums[j], nums[k]])
        return ans

    def binaray_search(self, nums: list[int], t: int,  j: int) -> int:
        n = len(nums)

        while j < n:
            m = (j + n) // 2
            if nums[m] < t:
                j = m + 1
            elif nums[m] > t:
                n = m
            else:
                return m
        return -1
```

```python
class Solution:
    # [-4, -1, -1, 0, 1, 2]
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        n = len(nums)
        ans = list()
        for i in range(n):
            if i != 0 and nums[i] == nums[i - 1]:
                continue
            j, k = i + 1, len(nums) - 1
            while j < k:
                s = nums[i] + nums[j] + nums[k]
                if s == 0:
                    ans.append([nums[i], nums[j], nums[k]])
                    j += 1
                    while j < k and nums[j] == nums[j - 1]:
                        j += 1
                    k -= 1
                    while j < k and nums[k] == nums[k + 1]:
                        k -= 1
                elif s > 0:
                    k -= 1
                else:
                    j += 1
        return ans
```
