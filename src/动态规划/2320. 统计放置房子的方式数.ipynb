```python
class Solution:
    def countHousePlacements(self, n: int) -> int:
        mod = 10**9 + 7

        @cache
        def dfs(i: int, left: bool, right: bool) -> int:
            if i == 0:
                return 1
            # 不放
            res = dfs(i - 1, True, True)
            if left:
                res += dfs(i - 1, False, True)
            if right:
                res += dfs(i - 1, True, False)
            if left and right:
                res += dfs(i - 1, False, False)
            return res % mod

        return dfs(n, True, True)
```
