```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        nums.sort()
        max_len = 0
        cur_len = 1
        for i in range(1, len(nums)):
            if nums[i - 1] + 1 == nums[i]:
                cur_len += 1
            elif nums[i - 1] == nums[i]:
                continue
            else:
                cur_len = 1
            max_len = max(max_len, cur_len)
        return max_len
```

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        if len(nums) == 0:
            return 0
        n_set = set(nums)
        num_min = min(nums)
        num_max = max(nums)
        cur_len = 0; max_len = 0
        for i in range(num_min, num_max + 1):
            if i in n_set:
                cur_len += 1
            else:
                cur_len = 0
            max_len = max(cur_len, max_len)
        return max_len
```

```python
class Solution:
    ## 如果x - 1在集合中，则不以x开头
    def longestConsecutive(self, nums: List[int]) -> int:
        nset = set(nums)

        ans = 0
        for num in nset:
            if num - 1 not in nset:
                cur = 1
                while num + 1 in nset:
                    cur += 1
                    num = num + 1
                ans = max(cur, ans)
        return ans
```
