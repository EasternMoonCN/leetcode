```python
from typing import List
```

```python
class Solution:
    def perfectMenu(self, materials: List[int], cookbooks: List[List[int]], attribute: List[List[int]], limit: int) -> int:
        # 收集食材条件下，饱腹感不小于limit的最大美味度
        n = len(cookbooks)
        def dfs(i: int, limit: int, path: int, m: List[int]):
            # 理论上，只要达到饱腹值，就是一种解，但是达到n时，能使美味值更大
            if i == n:
                # 达到饱腹值
                if limit <= 0:
                    return path
                # 没达到饱腹值
                return -1
            l1 = l2 = -1
            # 第i种料理的材料足够
            if all(x >= y for x, y in zip(m, cookbooks[i])):
                 # 制作第i种料理
                 # 下次还可以制作第i种料理(每个只能做一次)
                 l1 = dfs(i + 1, limit - attribute[i][1], path + attribute[i][0], [x - y for x, y in zip(m, cookbooks[i])])
            # 不制作第i种料理
            l2 = dfs(i + 1, limit, path, m)
            return max(l1, l2)
        return dfs(0, limit, 0, materials)
```
