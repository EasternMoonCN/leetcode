```python
from typing import List
from collections import deque
```

要么移最大值，要么移最小值
知道当前窗口内的最小值，寻找能够获取的最大值

```python
nums = [2,1,5]
nums.sort(reverse=True)
print(nums)
```

```python
class Solution:
    def minRemoval(self, nums: List[int], k: int) -> int:
        nums.sort()
        ideque = deque()
        ans = 0
        for num in nums:
            ideque.append(num)
            while ideque and ideque[0] * k < num:
                ideque.popleft()
            ans = max(ans, len(ideque))
        return len(nums) - ans
```

```python
class Solution:
    def minRemoval(self, nums: List[int], k: int) -> int:
        nums.sort()
        left = ans = 0
        for right, num in enumerate(nums):
            while nums[left] * k < num:
                left += 1
            ans = max(ans, right - left + 1)
        return len(nums) - ans
```
