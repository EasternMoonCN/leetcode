```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort()
        ans = []
        for x, y in intervals:
            if not ans:
                ans.append([x, y])
            else:
                x1, y1 = ans[-1]
                if y1 < x:
                    ans.append([x, y])
                else:
                    ans[-1][-1] = max(ans[-1][-1], y)
        return ans
```
