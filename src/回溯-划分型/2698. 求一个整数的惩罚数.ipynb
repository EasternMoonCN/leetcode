```python
class Solution:
    def punishmentNumber(self, n: int) -> int:
        # 划分字符串s
        def dfs(s: str, n: int, i: int, path: int):
            # 必须要等到长度满了，才算一次分割
            if i == len(s):
                return path == n
            for j in range(i, len(s)):
                # s[i:j + 1):分割，从i开始，到j+1(不包含)结束
                if dfs(s, n, j + 1, path + int(s[i:j + 1])):
                    # 只要存在一次分割满足，就直接返回
                    return True
            # 以i开头，任意长度的分割都不满足
            return False

        ans = 0
        for i in range(1, n + 1):
            if dfs(str(i * i), i, 0, 0):
                ans += i * i
        return ans
```
