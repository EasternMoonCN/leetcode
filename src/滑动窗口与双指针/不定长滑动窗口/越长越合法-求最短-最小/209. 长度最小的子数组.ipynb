```python
from typing import List
from math import inf
```

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        ans = inf
        res = left = 0
        for right, num in enumerate(nums):
            res += num
            # 找到一个不满足的
            while res >= target:
                ans = min(ans, right - left + 1)
                res -= nums[left]
                left += 1
        return ans if ans != inf else 0
```

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        n = len(nums)
        ans = n + 1
        res = left = 0
        for right, num in enumerate(nums):
            res += num
            # 循环到一个不满足之前
            while res - nums[left] >= target:
                res -= nums[left]
                left += 1
            if res >= target:
                ans = min(ans, right - left + 1)
        return ans if ans <= n else 0
```
