```python
from functools import cache
```

```python
class Solution:
    # 如何进行分割：
    # 1. 不同字符必须分割
    # 2.连续字符可分可不分
    # 3. 79最长连续4个，其他最长连续3个
    def countTexts(self, pressedKeys: str) -> int:
        n = len(pressedKeys)
        mod = 1000000007
        # dfs(i)：长度为i方案数
        @cache
        def dfs(i : int) -> int:
            if i < 0:
                return 1
            maxLen = 4 if pressedKeys[i] == '7' or pressedKeys[i] == '9' else 3
            r = 0
            # 便利所有长度
            for len in range(1, maxLen + 1):
                j = i - len + 1 # [j, i] 没有以i结尾，所有可能的j
                if j < 0:
                    break
                if pressedKeys[j] != pressedKeys[i]:
                    break
                r = (r + dfs(j - 1)) % mod # [j, i]有效，继续枚举前面dfs(j-1)
            return r
        return dfs(n - 1) % mod
```

```python
class Solution:
    # 如何进行分割：
    # 1. 不同字符必须分割
    # 2.连续字符可分可不分
    # 3. 79最长连续4个，其他最长连续3个
    def countTexts(self, pressedKeys: str) -> int:
        n = len(pressedKeys)
        mod = 1000000007
        dp = [1] + [0] * n
        for i in range(0, n):
            maxLen = 4 if pressedKeys[i] == '7' or pressedKeys[i] == '9' else 3
            # 必须要从后往前遍历，中间如果存在不连续的就跳出了
            for j in range(i, max(0, i - maxLen + 1) - 1, -1):
                if pressedKeys[j] != pressedKeys[i]:
                    break
                dp[i + 1] = (dp[i + 1] + dp[j]) % mod
        return dp[n] % mod
```
