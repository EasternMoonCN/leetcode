```python
from typing import List
from collections import defaultdict
```

```python
class Solution:
    def maxSubarrayLength(self, nums: List[int], k: int) -> int:
        idict = defaultdict(int)
        ans = left = 0
        for right, num in enumerate(nums):
            idict[num] += 1
            while idict[num] > k:
                out = nums[left]
                idict[out] -= 1
                left += 1
            ans = max(ans, right - left + 1)
        return ans
```
