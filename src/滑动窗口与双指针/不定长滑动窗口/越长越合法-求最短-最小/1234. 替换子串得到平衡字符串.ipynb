```python
from collections import Counter
```

```python
s = "abcdabcd"
cnt = Counter(s)
print(cnt)
print(cnt.values())
print(cnt.items())
print(cnt.keys())
print(Counter({"a":1, 'b':3}))
print(Counter({1, 2, 3, 4}))
```

```python
class Solution:
    # 保证窗口符之外的任意字符都不超过m次
    # 窗口之外不超过m，则一定能转换成平衡字符串
    def balancedString(self, s: str) -> int:
        n = len(s)
        # 每个字符m次
        m = n // 4
        # 字符串之外计数
        cnt = Counter(s)
        if len(cnt) == 4 and min(cnt.values()) == m:
            return 0
        ans = n + 1
        left = 0
        for right, c in enumerate(s):
            # 窗口之内:[left, right]
            cnt[c] -= 1
            # 满足窗口之外的任意字符都不超过m次
            while max(cnt.values()) <= m:
                ans = min(ans, right - left + 1)
                cnt[s[left]] += 1
                left += 1
        return ans
```
