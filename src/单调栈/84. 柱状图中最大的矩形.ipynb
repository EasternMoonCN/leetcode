```python
from typing import List
```

三次遍历
- 第一次求出左边比当前h小的
- 第二次求出右边比当前h小的
- 后面一次计算每个位置，和高度h

```python
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        n = len(heights)
        left = [-1] * n
        right = [n] * n
        stack = []
        for i, h in enumerate(heights):
            while stack and heights[stack[-1]] > h:
                right[stack.pop()] =  i
            stack.append(i)
        stack.clear()

        for i in range(n - 1, -1, -1):
            h = heights[i]
            while stack and heights[stack[-1]] > h:
                left[stack.pop()] =  i
            stack.append(i)

        max_val = 0
        for i, h in enumerate(heights):
            w = right[i] - left[i] - 1
            max_val = max(w * h, max_val)
        return max_val
```
