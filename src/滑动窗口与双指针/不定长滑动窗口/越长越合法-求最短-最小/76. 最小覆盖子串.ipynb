- 这题的思路在于，当right不断向右时，越会满足条件s覆盖t，但是当s覆盖t时，如何求最小呢？
- 此时只需要判断left向右，再次判断s是否覆盖t

- 如何判断s覆盖t呢
- 一种方法是统计t中各个元素的个数，需要的时候判断t中元素个数是否减到0，但是会有O(52)复杂度（集合中英文字母个数）
- 另一种方法是less表示t中不同元素个数，当t中元素个数减少到0时，就将less减1，当less等于0时，s也就包含t了，从而达到O(1)

```python
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        tdict = defaultdict(int)
        for c in t:
            tdict[c] += 1
        ans = ""
        left = 0

        for right, c in enumerate(s):
            if c in tdict:
                tdict[c] -= 1
            while self.softSubstr(tdict):
                tmp = s[left : right + 1]
                ans = tmp if ans == "" or len(tmp) < len(ans) else ans
                if s[left] in tdict:
                    tdict[s[left]] += 1
                left += 1
        return ans

    def softSubstr(self, tdict: dict) -> bool:
        for value in tdict.values():
            if value > 0:
                return False
        return True
```

```python
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        tdict = defaultdict(int)
        for c in t:
            tdict[c] += 1
        ans = ""
        left = 0
        # t中有多少个元素
        less = len(tdict)

        for right, c in enumerate(s):
            tdict[c] -= 1
            # 不再t中元素不会满足，在t中只有恰好s中有这么多才满足
            if tdict[c] == 0:
                less -= 1
            ## 只会受t中元素影响
            while less == 0:
                tmp = s[left : right + 1]
                ans = tmp if ans == "" or len(tmp) < len(ans) else ans
                # 先判断0保证了，left不在t中时，less不变，直到遇到left在t中时，less才改变
                if tdict[s[left]] == 0:
                    less += 1
                tdict[s[left]] += 1
                left += 1
        return ans

```
