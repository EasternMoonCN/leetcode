```python
from typing import List
```

```python
class Solution:
    def lower_bound(self, nums: List[int], target: int) -> int:
        # [l, h]
        l, h = 0, len(nums) - 1
        while l <= h:
            m = (l + h) // 2
            if nums[m] < target:
                l = m + 1
            else:   # nums[m] >= target
                h = m - 1
        return l

    def search(self, nums: List[int], target: int) -> int:
        pos = self.lower_bound(nums, target)
        if pos == len(nums) or pos < 0:
            return -1
        return pos if nums[pos] == target else -1
```
