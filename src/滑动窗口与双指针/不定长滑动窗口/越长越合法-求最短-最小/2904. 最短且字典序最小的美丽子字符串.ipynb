```python
class Solution:
    def shortestBeautifulSubstring(self, s: str, k: int) -> str:
        ans = ""
        left = res = 0
        for right, c in enumerate(s):
            if c == '1':
                res += 1
            # s[left] == '0'：去掉左侧多余0
            while res > k or (left <= right and s[left] == '0'):
                if s[left] == '1':
                    res -= 1
                left += 1
            if res == k:
                if len(ans) == 0:
                    ans = s[left : right + 1]
                # 比较长度
                elif len(ans) < (right - left + 1):
                    ans = ans
                elif len(ans) > (right - left + 1):
                    ans = s[left : right + 1]
                # 比较字典序
                elif ans < s[left : right + 1]:
                    ans = ans
                else:
                    ans = s[left : right + 1]
        return ans

```

```python
class Solution:
    def shortestBeautifulSubstring(self, s: str, k: int) -> str:
        # s中字符'1'个数小于k，不满足条件(所以left <= right不需要了)
        if s.count('1') < k:
            return ""
        # 个数大于等于k情况下，一定有串
        ans = ""
        left = res = 0
        for right, c in enumerate(s):
            res += int(c)
            # s[left] == '0'：去掉左侧多余0
            while res > k or s[left] == '0':
                res -= int(s[left])
                left += 1
            if res == k:
                t = s[left : right + 1]
                if len(ans) == 0 or len(t) < len(ans) or (len(t) == len(ans) and t < ans):
                    ans = t
        return ans

```
