```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        l1, l2 = [], []
        def dfs(head: Optional[ListNode]) -> None:
            if head is None:
                return
            l1.append(head.val)
            dfs(head.next)
            l2.append(head.val)
        dfs(head)
        return l1 == l2
```
