```python
from typing import List
from functools import reduce
import operator
class Solution:
    # O（2^n）
    def checkEqualPartitions(self, nums: List[int], target: int) -> bool:
        total = reduce(operator.mul, nums)
        if target * target != total:
            return False
        n = len(nums)
        # 选一些元素，乘积为target
        def dfs(i, m):
            if i == n:
                return m == target
            # 选    # 不选
            return dfs(i + 1, m * nums[i]) or dfs(i + 1, m)
        return dfs(0, 1)
```

```python
import math
class Solution:
    def checkEqualPartitions(self, nums: List[int], target: int) -> bool:
        # total = reduce(operator.mul, nums)
        total = math.prod(nums)
        if target * target != total:
            return False
        n = len(nums)
        # 选哪元素，乘积为target
        def dfs(i, m):
            if m == target:
                return True
            for j in range(i, n):
                if dfs(j + 1, m * nums[j]):
                    return True
            return False
        return dfs(0, 1)
```
