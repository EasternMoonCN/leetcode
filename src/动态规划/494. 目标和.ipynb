```python
from typing import List
from functools import cache
```

```python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        if len(nums) == 1:
            return 1
        # 挑选一些数，其和为t
        # 因为t在sum时加了2遍，所以要//2
        # （正 + 负）- target
        # 正 - target = 负
        t = (sum(nums) - target) // 2
        n = len(nums)
        @cache
        def dfs(i: int, t: int) -> int:
            if i < 0:
                return t == 0
            r1 = dfs(i - 1, t - nums[i])
            r2 = dfs(i - 1, t)
            return r1 + r2
        return dfs(n - 1, t)
```

```python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        # p（正数），s（和），s-p（负数）
        # p - （s-p） = t
        # 2p = t + s
        target = sum(nums) + target
        # nums[i]都是正数，无论怎么选，都是非负数
        if target < 0 or target % 2 != 0:
            return 0
        target = target // 2
        n = len(nums)
        # dp[i][t]：选前i个数，其和为t的方案数
        dp = [0] * (target + 1)
        dp[0] = 1
        for i in range(n):
            # t 依赖 t - nums[i]前面的值，所以要反着算
            for t in range(target, -1, -1):
                if t >= nums[i]:
                    dp[t] = dp[t - nums[i]] + dp[t]
                else:
                    dp[t] = dp[t]
        return dp[target]
```
