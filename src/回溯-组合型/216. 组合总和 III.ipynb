```python
class Solution:
    def combinationSum3(self, k: int, n: int) -> List[List[int]]:
        ans = []
        path = []
        def dfs(i, n):
            if len(path) == k and n == 0:
                ans.append(path.copy())
                return
            if i > 9 or n < 0 or len(path) > k:
                return
            if i > n:
                return
            path.append(i)
            dfs(i + 1, n - i)
            path.pop()
            dfs(i + 1, n)
        dfs(1, n)
        return ans
```

```python
class Solution:
    def combinationSum3(self, k: int, n: int) -> List[List[int]]:
        ans = []
        path = []
        # O(k*C(9, k))
        def dfs(i):
            nonlocal n
            if n == 0 and len(path) == k:
                ans.append(path.copy())
                return
            for j in range(i, 10):
                if n < 0 or len(path) >= k or j > n:
                    return
                path.append(j)
                n -= j
                dfs(j + 1)
                n += j
                path.pop()
        dfs(1)
        return ans
```

```python
class Solution:
    def combinationSum3(self, k: int, n: int) -> List[List[int]]:
        ans = []
        path = []
        # O(k*C(9, k))
        def dfs(i, t):
            d = k - len(path) # 还需要选d个数
            if t < 0 or t > (i + i - d + 1) * d // 2:
                return
            if d == 0:
                ans.append(path.copy())
                return
            for j in range(i, d - 1, - 1):
                path.append(j)
                dfs(j - 1, t - j)
                path.pop()
        dfs(9, n)
        return ans
```
