```python
class Solution:
    ans = inf
    # 如果开过锁，剑的能量+k
    # 如果没没过锁，剑的能量+x
    def findMinimumTime(self, strength: List[int], k: int) -> int:
        n = len(strength)
        # x: 当前能量x
        def dfs(i: int, p: int, x: int, nums: List[int]) -> None:
            # 当前花费的时间，以及超过已有的答案
            if i - 1 >= self.ans:
                return
            if len(nums) == 0:
                self.ans = min(self.ans, i - 1)
                return
            # 记录本次是否可选
            flag = False
            for j, num in enumerate(nums):
                if p < num:
                    continue
                flag = True
                # 当前能量满足
                dfs(i + 1, x + k, x + k, nums[:j] + nums[j + 1:])

            # 当前能量无法满足
            if not flag:
                dfs(i + 1, p + x, x, nums)
        dfs(0, 0, 1, strength)
        return self.ans
```

```python
class Solution:
    ans = inf
    # 如果开过锁，剑的能量+k
    # 如果没没过锁，剑的能量+x
    def findMinimumTime(self, strength: List[int], k: int) -> int:
        n = len(strength)
        selected = [False] * n
        # i：开了几把锁，t：花费时间
        def dfs(i: int, t: int) -> None:
            if t >= self.ans:
                return
            if i == n:
                self.ans = min(self.ans, t)
                return
            # 当前的增加因子
            x = 1 + i * k
            for j in range(0, n):
                if selected[j]:
                    continue
                selected[j] = True
                # t：开当前这把锁花费的时间
                # p = 0 + t * x (经历了t轮什么也不做)
                # t * x >= strength[j]
                # t >= strength[j] / x （向上取整）
                dfs(i + 1, t + (strength[j] - 1) // x + 1)
                selected[j] = False
        dfs(0, 0)
        return self.ans
```
