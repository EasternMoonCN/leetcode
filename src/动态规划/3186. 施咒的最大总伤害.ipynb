```python
class Solution:
    def maximumTotalDamage(self, power: List[int]) -> int:
        cnt = dict(Counter(power))
        a = list(cnt)
        a.sort()
        n = len(cnt)
        dp = [0] * (n + 2)
        dp[2] = a[0] * cnt[a[0]]
        for i in range(1, n):
            if a[i] - 1  == a[i - 1] and a[i] - 2 == a[i - 2]:
                dp[i + 2] = max(dp[i + 1], cnt[a[i]] * a[i] + dp[i - 1])
            elif a[i] - 1  == a[i - 1] or a[i] - 2 == a[i - 1]:
                dp[i + 2] = max(dp[i + 1], cnt[a[i]] * a[i] + dp[i])
            else:
                dp[i + 2] = cnt[a[i]] * a[i] + dp[i + 1]
        return dp[n + 1]
```

```python
class Solution:
    def maximumTotalDamage(self, power: List[int]) -> int:
        cnt = dict(Counter(power))
        nums = sorted(list(Counter(power)))
        n = len(nums)
        # dp[i][j]: 第i种咒语，j=0选，j=1跳过一次，j=2跳过两次
        dp = [[0] * 3 for _ in range(n + 1)]
        for i in range(n):
            # 选前面nums[k] < nums[i] - 2的
            k = i
            while k >= 0:
                if nums[k] < nums[i] - 2:
                    break
                k -= 1
            best = max(dp[k + 1])
            dp[i + 1][0] = cnt[nums[i]] * nums[i] + best
            dp[i + 1][1] = dp[i][0] # 第一次跳过，那它由前面选来的
            dp[i + 1][2] = dp[i][1] # 第二次跳过，由前面第一次跳过来的
        return max(dp[n])
```

```python
class Solution:
    def maximumTotalDamage(self, power: List[int]) -> int:
        cnt = dict(Counter(power))
        nums = sorted(cnt)
        n = len(cnt)
        dp = [0] * (n + 1)
        # 记录小于nums[i] - 2 的下一个位置
        j = 0
        for i in range(n):
            while nums[j] < nums[i] - 2:
                j += 1
            dp[i + 1] = max(dp[i], nums[i] * cnt[nums[i]] + dp[j]) # 上一个位置
        return dp[n]
```
