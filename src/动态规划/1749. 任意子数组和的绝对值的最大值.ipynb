```python
class Solution:
    def maxAbsoluteSum(self, nums: List[int]) -> int:
        n = len(nums)
        pre_max = pre_min = 0
        max_val = -inf
        for i, x in enumerate(nums):
            pre_max = max(x, pre_max + x)
            pre_min = min(x, pre_min + x)
            max_val = max(max_val, max(pre_max, abs(pre_min)))
        return max_val
```

```python
class Solution:
    def maxAbsoluteSum(self, nums: List[int]) -> int:
        pre = 0
        min_pre = max_pre = 0
        # 前缀和种最大值 - 最小值，最大值在最小值右边，相当于最大子数组和，最大值在最小值左边，相当于最小子数组和
        for i, x in enumerate(nums):
            pre += x
            min_pre = min(min_pre, pre)
            max_pre = max(max_pre, pre)
        return max_pre - min_pre
```
