```python
from typing import List
class Solution:
    def subsetXORSum(self, nums: List[int]) -> int:
        ans = [0]
        path = []
        n = len(nums)
        def xor(path):
            if len(path) == 0:
                return 0
            elif len(path) == 1:
                return path[0]
            ans = 0
            for num in path:
                ans ^= num
            return ans

        def dfs(i):
            ans[0] = ans[0] + xor(path)
            for j in range(i, n):
                path.append(nums[j])
                dfs(j + 1)
                path.pop()
        dfs(0)
        return ans[0]
```

```python
import operator
from functools import reduce
class Solution:
    def subsetXORSum(self, nums: List[int]) -> int:
        ans = [0]
        path = []
        n = len(nums)

        def dfs(i):
            ans[0] = ans[0] + reduce(operator.xor, path) if path else 0
            for j in range(i, n):
                path.append(nums[j])
                dfs(j + 1)
                path.pop()
        dfs(0)
        return ans[0]
```

```python
class Solution:
    def subsetXORSum(self, nums: List[int]) -> int:
        n = len(nums)
        def dfs(i, path):
            total = reduce(operator.xor, path) if path else 0 # 每个节点都是一个完整集合
            for j in range(i, n):
                total += dfs(j + 1, path + [nums[j]])
            return total
        return dfs(0, [])
```

```python
class Solution:
    def subsetXORSum(self, nums: List[int]) -> int:
        n = len(nums)
        def dfs(i, path):
            if i == n: # 只有到达叶子节点，才是一个完整集合
                return reduce(operator.xor, path) if path else 0
            # 选
            total = dfs(i + 1, path + [nums[i]])
            # 不选
            total += dfs(i + 1, path)
            return total
        return dfs(0, [])
```
