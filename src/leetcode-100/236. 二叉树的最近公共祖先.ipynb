```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    # 先序，分类讨论
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        # 1. 当前节点空，左节点，右节点，直接返回
        if root is None or root is p or root is q:
            return root
        # 2. 当前节点左子树（
            # 1. 左右子树都有，返回当前节点，
            # 2.只有左子树有，返回左子树，（这两种不需要再递归下去了）
            # 3.只有右子树右，返回右子树，
            # 4.左右都没有）
        left = self.lowestCommonAncestor(root.left, p, q)
        right = self.lowestCommonAncestor(root.right, p, q)
        if left and right:
            return root
        if left:
            return left
        return right    # 包含左右都没有情况
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    # 倒序，左右都有，或左右存在一个，且当前是 就 更新
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        ans = [None]
        # 判断root是否p或q
        def dfs(root):
            if root is None:
                return False
            left = dfs(root.left)
            right = dfs(root.right)
            if (left and right) or ((left or right) and (root is p or root is q)):
                ans[0] = root
            return left or right or root is p or root is q
        dfs(root)
        return ans[0]
```
