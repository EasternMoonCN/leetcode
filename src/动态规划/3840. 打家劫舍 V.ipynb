```python
class Solution:
    def rob(self, nums: List[int], colors: List[int]) -> int:
        n = len(colors)
        dp = [0] * (n + 1)
        dp[1] = nums[0]
        for i in range(1, n):
            if colors[i] != colors[i - 1]:      # 颜色不同，选前一个
                dp[i + 1] = nums[i] + dp[i]
            else:                               # 颜色相同，隔一个
                dp[i + 1] = nums[i] + dp[i - 1]
            dp[i + 1] = max(dp[i + 1], dp[i])   # 选或不选，由于dp[i] + nums[i] 一定比dp[i]大，所以可以合并到颜色相同
        return dp[n]
```
