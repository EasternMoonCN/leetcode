```python
customers = [1,0,1,2,1,1,7,5]
grumpy = [0,1,0,1,0,1,0,1]
print(customers * 2)
print(sum(c if c >= 0 else 0 for c in customers))
```

```python
from typing import List
```

```python
class Solution:
    # O(N)
    def maxSatisfied(self, customers: List[int], grumpy: List[int], minutes: int) -> int:
        # 不生气时对应的顾客数量
        nums = [c * g for c, g in zip(customers, grumpy)]
        print(nums)
        ans = res = 0
        for i, num in enumerate(nums):
            res += num
            left = i - minutes + 1
            if left < 0:
                continue
            ans = max(ans, res)
            res -= nums[left]
        return ans + sum(c if g == 0 else 0 for c, g in zip(customers, grumpy))
```
