```python
from typing import List
```

```python
string = "s"
print(type(string))
print(ord(string))
print(chr(ord(string)))
print(str(ord(string)))
```

```python
class Solution:
    # >= f(x)
    # > f(x) ==> >= f(x + 1)
    def lower_bound(self, letters: List[str], target: chr) -> int:
        l, h = 0, len(letters) - 1
        while l <= h:
            m = (l + h) // 2
            if letters[m] < target:
                l = m + 1
            else:
                h = m - 1
        return l

    def nextGreatestLetter(self, letters: List[str], target: str) -> str:
        pos = self.lower_bound(letters, chr(ord(target) + 1))
        if pos == len(letters):
            return letters[0]
        return letters[pos]
```

```python
if __name__ == '__main__':
    letters = ["c","f","j"]
    target = "c"
    solution = Solution()
    print(solution.nextGreatestLetter(letters, target))
```
