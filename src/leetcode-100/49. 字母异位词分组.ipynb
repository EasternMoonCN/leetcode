```python
class Solution:
    ## 以字符串二进制相同作为key，无法解决长度一致情况下，字符串重复下，不同的也会想等
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        # 二进制 -> str
        s_dict = defaultdict(list)
        for s in strs:
            i = 0
            for c in s:
                i += 1 << (ord(c) - ord('a'))
            s_dict[i, len(s)].append(s)
        result = []
        for _, r in s_dict.items():
            result.append(r)
        return result
```

```python
class Solution:
    ## 排序之后的字符串作为key
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        s_dict = defaultdict(list)
        for s in strs:
            s_dict["".join(sorted(s))].append(s)
        result = []
        for r in s_dict.values():
            result.append(r)
        return result
```
