```python
from  typing import List
import bisect
```

```python
potions = [8,5,8]
potions.sort()
print(potions)
```

```python
class Solution:
    # O((n+m)log(m))
    def successfulPairs(self, spells: List[int], potions: List[int], success: int) -> List[int]:
        ## mlog(m)
        potions.sort()
        res = list()
        ## nlog(m)
        for spell in spells:
            # 向上取证
            target = (success + spell - 1) // spell
            # [pos, len(potions) - 1] 都满足
            pos = bisect.bisect_left(potions, target)
            res.append(len(potions) - pos)
        return res
```
