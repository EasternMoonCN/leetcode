```python
from typing import List
```

```python
class Solution:
    def longestSubarray(self, nums: List[int]) -> int:
        cnt = ans = left = 0
        for right, num in enumerate(nums):
            while num == 0 and cnt == 1:
                if nums[left] == 0:
                    cnt -= 1
                left += 1
            if num == 0:
                cnt += 1
            # 因为必须删一个元素，所以-1
            ans = max(ans, right - left + 1 - 1)
        return ans
```

```python
class Solution:
    # O(N)
    def longestSubarray(self, nums: List[int]) -> int:
        cnt = ans = left = 0
        for right, num in enumerate(nums):
            # 先入队列
            cnt += 1 - num
            while cnt > 1:
                cnt -= 1 - nums[left]
                left += 1
            # 因为必须删一个元素，所以-1
            ans = max(ans, right - left + 1 - 1)
        return ans
```
