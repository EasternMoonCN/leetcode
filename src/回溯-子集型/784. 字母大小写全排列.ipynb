```python
class Solution:
    def letterCasePermutation(self, s: str) -> List[str]:
        n = len(s)
        ans = []
        def dfs(i, path):
            if i == n:
                ans.append("".join(path))
                return
            c = s[i]
            # 转
            if c.isalpha():
                dfs(i + 1, path + [c.upper() if c.islower() else c.lower()])
            # 不转
            dfs(i + 1, path + [s[i]])
        dfs(0, [])
        return ans
```

```python
class Solution:
    def letterCasePermutation(self, s: str) -> List[str]:
        n = len(s)
        ans = []
        # 当前位置，有哪些可选
        def dfs(i, path):
            if i == n:
                ans.append("".join(path))
                return
            # 构造可选字符
            cset = []
            if s[i].isalpha():
                cset.append(s[i].lower() if s[i].isupper() else s[i].upper())
            cset.append(s[i])
            for c in cset:
                dfs(i + 1, path + [c])
        dfs(0, [])
        return ans
```
