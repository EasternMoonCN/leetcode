```python
class Solution:
    def minimumCoins(self, prices: List[int]) -> int:
       n = len(prices)
       # n
       half = (n + 1) // 2
       mem = [-1] * half
       # dfs(i): 购买i及之后的水果花费的最少金币
       def dfs(i: int) -> int:
           # 超过n个，说明都免费了，不需要花钱
           if i >= n:
               return 0
           if i >= half:
               return prices[i]
           if mem[i] != -1:
               return mem[i]
           # 购买i后，[i+1, i+i+1]免费，i+i+2不免费，结尾为i+i+3
           # 免费体现在中间的可以不选，当j在[i + 1, i + i + 2]之间选择，j需要付费，[i + 1, j - 1]之间免费
           tmp = inf
           for j in range(i + 1, i + i + 3):
               tmp = min(tmp, dfs(j))
           mem[i] = prices[i] + tmp
           return mem[i]
       res = dfs(0)
       return res
```

```python
class Solution:
    def minimumCoins(self, prices: List[int]) -> int:
       n = len(prices)
       # n
       half = (n +1) // 2
       dp = [0] * half
       for i in range(half - 1, -1, -1):
           # print(f"{i + 1}, {i + i + 3}")
           tmp = inf
           for j in range(i + 1, i + i + 3):
               if j >= n:
                   tmp = 0
                   break
               # prices[j]不一定是[i + 1, i + i + 2]之间的最小值，所以还需要遍历
               elif j >= half:
                   tmp = min(tmp, prices[j])
               else:
                   tmp = min(tmp, dp[j])
           dp[i] = tmp + prices[i]
       return dp[0]
```
