```python
from typing import List
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        cache = {}
        def dfs(i: int) -> int:
            if i in cache:
                return cache[i]
            if i >= n:
                return 0
            # 偷i
            r1 = dfs(i + 2) + nums[i]
            # 不偷i
            r2 = dfs(i + 1)
            cache[i] = max(r1, r2)
            return cache[i]
        return dfs(0)
```

```python
class Solution:
    ans = 0
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        def dfs(i: int, path: int) -> None:
            if i >= n:
                self.ans = max(self.ans, path)
                return
            # 偷i
            dfs(i + 2, path + nums[i])
            # 不偷i
            dfs(i + 1, path)
        dfs(0, 0)
        return self.ans
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        cache = {}
        def dfs(i: int) -> int:
            if i in cache:
                return cache[i]
            if i < 0:
                return 0
            # 偷i
            r1 = dfs(i - 2) + nums[i]
            # 不偷i
            r2 = dfs(i - 1)
            cache[i] = max(r1, r2)
            return cache[i]
        return dfs(n - 1)
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        # 子问题dp[i]：从【前】i个（不包含i）房子的最大价值
        # 不选：从【前】i-1个房子中得到的最大价值
        # 选：从【前】i-2个房子中得到的最大价值
        # 当前操作：【第】i个房子选不选
        dp = [0] * (n + 2)
        for i in range(n):
            dp[i + 2] =  max(dp[i] + nums[i], dp[i + 1])
        return dp[n + 1]
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        f1 = f2 = c = 0
        for i in range(n):
            c =  max(f1 + nums[i], f2)
            f1 = f2
            f2 = c
        return c
```
