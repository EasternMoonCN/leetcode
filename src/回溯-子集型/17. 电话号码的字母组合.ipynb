```python
from collections import defaultdict
from typing import List
class Solution:
    def __init__ (self):
        self.d2c = defaultdict(list)
        c = 'a'
        for i in range(2, 10):
            n = 4 if i == 7 or i == 9 else 3
            for j in range(n):
                self.d2c[str(i)].append(c)
                c = chr(ord(c) + 1)

    # 原问题：构造长为n的字符串
    # 子问题：构造长为n-1多字符串
    # 每个一层最多4，枚举n层，时间复杂度O(4^n)
    def letterCombinations(self, digits: str) -> List[str]:
        n = len(digits)
        ans = []
        def dfs(i: int, tmp: str, ans: List[str]):
            if i == n:
                ans.append(tmp)
                return
            # 选哪个
            for c in self.d2c[digits[i]]:
                tmp += c
                dfs(i + 1, tmp, ans)
                tmp = tmp[:-1]
        dfs(0, "", ans)
        return ans

```

```python
class Solution:
    mapping = ['', '', 'abc', 'def', 'ghi', 'jkl', 'mno', 'pqrs', 'tuv', 'wxyz']

    def letterCombinations(self, digits: str) -> List[str]:
        n = len(digits)
        ans = []
        path = [''] * n
        def dfs(i: int):
            if i == n:
                ans.append("".join(path))
                return
            # 选哪个
            for c in self.mapping[int(digits[i])]:
                # path.append(c)
                path[i] = c # 第i此都会被不同的值覆盖
                dfs(i + 1)
        dfs(0)
        return ans

```
