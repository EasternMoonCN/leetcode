```python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        ## 不可以用滑动窗口，因为nums存在负数，s大于k的情况下，不一定要减少窗口
        for right, num in enumerate(nums):
            s += num
            while s > k:
                s -= nums[left]
                left += 1
            if s == k and left <= right:
                ans += 1
        return ans
```

暴力做法

```python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        ans = 0
        for i in range(len(nums)):
            s = 0
            for j in range(i , len(nums)):
                s += nums[j]
                if s == k:
                    ans += 1
        return ans
```

前缀和+字典

```python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        n = len(nums)
        pre = [0] * (n + 1)
        # 前缀和
        for i in range(n):
            pre[i + 1] = pre[i] + nums[i]

        ans = 0
        p_dict = defaultdict(int)
        for p in pre:
            # p = [0, i - 1]，右边
            # 看字典里面有没有左边
            ans += p_dict[p - k]
            p_dict[p] += 1

        return ans
```

```python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        s = ans = 0
        cnt = defaultdict(int)
        cnt[0] = 1  # 空前缀元素和=0
        for num in nums:
            s += num # 1 2
            ans += cnt[s - k] # 0 1
            cnt[s] += 1   # [1:1, 2:1]
        return ans
```
