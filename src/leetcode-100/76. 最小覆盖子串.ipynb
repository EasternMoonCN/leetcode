```python
class Solution:
    # 首先找到一个窗口包含子串t，然后不断缩小窗口，直到不包含子串
    def minWindow(self, s: str, t: str) -> str:
        cnt = defaultdict(int)
        for c in t:
            cnt[c] -= 1
        # t中有多少不同的字符
        kinds = len(cnt)
        left = 0
        index, length = -1, len(s) + 1
        for right, c in enumerate(s):
            cnt[c] += 1
            # 只有c存在于t中，才会=0
            if cnt[c] == 0:
                kinds -= 1
            while kinds == 0:
                if right - left + 1 < length:
                    index = left
                    length = right - left + 1
                # 什么情况下，kinds需要+1？
                # 当元素个数为0
                if cnt[s[left]] == 0:
                    kinds += 1
                # 为什么-1放后面？
                # 在t中情况： 窗口中s[left]字母和t中一样，==0，-1
                #             窗口中s[left]字母大于t，>0，此轮不管
                # 不在t中情况：窗口中s[left]字母不在tt中，>0，不管
                cnt[s[left]] -= 1
                left += 1
        return "" if index == -1 else s[index : index + length]
```
