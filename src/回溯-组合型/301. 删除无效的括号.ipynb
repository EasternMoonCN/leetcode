```python
class Solution:
    def removeInvalidParentheses(self, s: str) -> List[str]:
        ans = set()
        length = 0
        n = len(s)
        def dfs(i: int, left: int, path: List[str]) -> None:
            nonlocal length
            if i == n:
                if left ==0:
                    if len(path) > length:
                        ans.clear()
                        ans.add("".join(path))
                        length = len(path)
                    elif len(path) == length:
                        ans.add("".join(path))
                return
            isLeft = (s[i] == "(")
            isAlpha = s[i].isalpha()

            # 字符不删/括号删
            if isAlpha:
                dfs(i + 1, left, path + [s[i]])
            else:
                dfs(i + 1, left, path)

            # 括号不删
            if not isAlpha and (isLeft or left > 0):
                dfs(i + 1, left + 1 if isLeft else left - 1, path + [s[i]])
        dfs(0, 0, [])
        return list(ans)
```

```python
class Solution:
    def removeInvalidParentheses(self, s: str) -> List[str]:
        ans = []
        n = len(s)
        def dfs(i: int, left: int, path: List[str]) -> None:
            # 剩余字符无法达到最优
            if ans and len(path) + (n - i) < len(ans[0]):
                return
            if i == n:
                if left == 0:
                    c = "".join(path)
                    if ans:
                        if len(c) > len(ans[0]):
                            ans.clear()
                            ans.append(c)
                        elif len(c) == len(ans[0]) and c not in ans:
                            ans.append(c)
                    else:
                        ans.append(c)
                return
            # 字符串不能删
            if s[i].isalpha():
                dfs(i + 1, left, path + [s[i]])
            else:
                isLeft = (s[i] == "(")
                # 删括号
                dfs(i + 1, left, path)
                # 不删括号
                if isLeft or left > 0:
                    dfs(i + 1, left + 1 if isLeft else left - 1, path + [s[i]])
        dfs(0, 0, [])
        return ans
```
