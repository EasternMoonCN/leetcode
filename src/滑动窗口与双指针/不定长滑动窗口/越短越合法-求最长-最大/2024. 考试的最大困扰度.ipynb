```python
from collections import defaultdict
```

```python
class Solution:
    def maxConsecutiveAnswersKey(self, answerKey: str, k: int, key: chr) -> int:
        left = ans = 0
        for right, answer in enumerate(answerKey):
            k = k - 1 if answerKey[right] == key else k
            while k < 0:
                k = k + 1 if answerKey[left] == key else k
                left += 1
            ans = max(ans, right - left + 1)
        return ans

    def maxConsecutiveAnswers(self, answerKey: str, k: int) -> int:
        return max(self.maxConsecutiveAnswersKey(answerKey, k, 'T'), self.maxConsecutiveAnswersKey(answerKey, k, 'F'))
```

```python
class Solution:
    # 窗口内至多k个T 或 F
    def maxConsecutiveAnswers(self, answerKey: str, k: int) -> int:
        idict = defaultdict(int)
        left = ans = 0
        for right, c in enumerate(answerKey):
            idict[c] += 1
            # 为什么是T和F都超过k才进入？
            # 在T 或 F没有超过情况下，下面max条件一定有一个满足
            # T 和 F都不满足情况下，left会一直加到有一个满足，表示[left, right]中至多不超过K个T或F
            while idict['T'] > k and idict['F'] > k:
                idict[answerKey[left]] -= 1
                left += 1
            ans = max(ans, right - left + 1)
        return ans
```
