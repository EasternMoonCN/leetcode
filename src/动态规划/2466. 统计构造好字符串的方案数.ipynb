```python
from functools import cache
from typing import List
```

```python
class Solution:
    # 最多可以拼接zero+one长度的字符
    # 长度字符需要满足[low, higt]
    def countGoodStrings(self, low: int, high: int, zero: int, one: int) -> int:
        # 构建长度为[low, higt]的字符
        ans = []
        def dfs(path: List[str]) -> None:
            n = len(path)
            if n > high:
                return
            if low <= n <= high:
                ans.append("".join(path))
            # 选0
            dfs( path + ['0'] * zero)
            # 选1
            dfs(path + ['1'] * one)
        dfs([])
        ## print(ans)
        return len(ans) % (10**9 + 1)
```

```python
class Solution:
    def countGoodStrings(self, low: int, high: int, zero: int, one: int) -> int:
        # 构建字符传长度为i的方案数
        @cache
        def dfs(i) -> int:
            if i <= 0:
                return 1 if i == 0 else 0
            # 选0
            return dfs(i - zero) + dfs(i - one)
        cnt = 0
        for i in range(low, high + 1):
            cnt += dfs(i)
        return cnt % (10**9 + 7)
```

```python
class Solution:
    def countGoodStrings(self, low: int, high: int, zero: int, one: int) -> int:
        mod = 1000000007
        # dp = [0] * (high + 1)
        # dp[0] = 1
        dp = [1] + [0] * high
        ans = 0
        for i in range(high + 1):
            if i >= zero:
                dp[i] = dp[i - zero]
            if i >= one:
                dp[i] = (dp[i] + dp[i - one]) % mod
            if i >= low:
                ans = (ans + dp[i]) % mod
        return ans
```
