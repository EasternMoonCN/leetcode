给你字符串 s 和整数 k 。

请返回字符串 s 中长度为 k 的单个子字符串中可能包含的最大元音字母数。

英文中的 元音字母 为（a, e, i, o, u）。



示例 1：
- 输入：s = "abciiidef", k = 3
- 输出：3
- 解释：子字符串 "iii" 包含 3 个元音字母。

示例 2：
- 输入：s = "aeiou", k = 2
- 输出：2
- 解释：任意长度为 2 的子字符串都包含 2 个元音字母。

示例 3：
- 输入：s = "leetcode", k = 3
- 输出：2
- 解释："lee"、"eet" 和 "ode" 都包含 2 个元音字母。

示例 4：
- 输入：s = "rhythms", k = 4
- 输出：0
- 解释：字符串 s 中不含任何元音字母。

示例 5：
- 输入：s = "tryhard", k = 4
- 输出：1

```python
vowel = ['a', 'e', 'i', 'o', 'u']
print('a' in vowel)
print('b' in vowel)
```

```python
from collections import deque
```

```python
class Solution:
    def maxVowels(self, s: str, k: int) -> int:
        vowel = ['a', 'e', 'i', 'o', 'u']
        arr = deque()
        ans = 0
        maxAns = 0
        for c in s:
            arr.append(c)
            if len(arr) > k:
                tmp = arr.popleft()
                if tmp in vowel:
                    ans -= 1
            if c in vowel:
                ans += 1
                maxAns = max(maxAns, ans)
        return maxAns
```

```python
class Solution:
    # 时间复杂度O(N)，每个元素遍历一次
    def maxVowels(self, s: str, k: int) -> int:
        vowel = ['a', 'e', 'i', 'o', 'u']
        left = 0
        ans = 0
        maxAns = 0
        for right, c in enumerate(s):
            # 保证有k-1个空间，当满了k个空间，需要移除一个
            if right - left + 1 > k:
                # 移除的元素是否是元音
                if s[left] in vowel:
                    ans -= 1
                left += 1
            # 新增加的元素是否是元音
            if c in vowel:
                ans += 1
                maxAns = max(maxAns, ans)
        return maxAns
```
