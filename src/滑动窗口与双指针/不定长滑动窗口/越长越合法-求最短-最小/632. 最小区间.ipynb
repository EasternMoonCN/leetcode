```python
class Solution:
    def smallestRange(self, nums: List[List[int]]) -> List[int]:
        min_v = nums[0][0]
        max_v = nums[0][-1]
        n = len(nums)
        for i in range(1, n):
            min_v = min(min_v, nums[i][0])
            max_v = max(max_v, nums[i][-1])

        lcount = [set() for i in range(n)]
        for i in range(n):
            for j in range(len(nums[i])):
                lcount[i].add(nums[i][j])

        left = min_v
        less = [1] * n

        def is_full(less :List[int]) -> bool:
            for l in less:
                if l > 0:
                    return False
            return True

        res_l, res_r = min_v, max_v
        for right in range(min_v, max_v + 1):
            for i, l in enumerate(lcount):
                if right in l:
                    less[i] -= 1
            while is_full(less):
                if res_r - res_l > right - left or (res_r - res_l == right - left and res_l > min_v):
                    res_l = left
                    res_r = right
                for i, l in enumerate(lcount):
                    if left in l:
                        less[i] += 1
                left += 1
        return [res_l, res_r]
```

```python
class Solution:
    # 第一个合法区间就是每个列表首个元素之间的最小值，最大值
    def smallestRange(self, nums: List[List[int]]) -> List[int]:
        # 每个列表第一个元素，第i个列表，位置
        # heap = [(arr[0], i, 0) for i, arr in enumerate(nums)]
        heap = []
        for i, arr in enumerate(nums):
            heapq.heappush(heap, (arr[0], i, 0))
        # 第一个合法区间左端点
        res_l = heap[0][0]
        # 第一个合法区间右端点
        res_r = r = max(arr[0] for arr in nums)
        # 堆顶元素还有下一个元素
        while heap[0][2] + 1 < len(nums[heap[0][1]]):
            _, i, j = heap[0]
            # 堆顶顶下一个元素
            x = nums[i][j + 1]
            heapq.heapreplace(heap, (x, i, j + 1))
            # 更新合法区间左端点
            l = heap[0][0]
            # r原先已经是列表首个元素最大值了，最需和新进元素比较
            r = max(r, x)
            if r - l < res_r - res_l:
                res_l = l
                res_r = r
        return [res_l, res_r]
```

```python
class Solution:
    def smallestRange(self, nums: List[List[int]]) -> List[int]:
        pairs = sorted((x, i) for i, arr in enumerate(nums) for x in arr)
        ans_l, ans_r = -inf, inf
        empty = len(nums)
        cnt = [0] * empty
        left = 0
        for r, i in pairs:
            if cnt[i] == 0:
                empty -= 1
            cnt[i] += 1
            while empty == 0:
                l, i = pairs[left]
                if r - l < ans_r - ans_l:
                    ans_l, ans_r = l, r
                cnt[i] -= 1
                if cnt[i] == 0: # 不包含nums[i] 列表中数字
                    empty += 1
                left += 1
        return [ans_l, ans_r]
```
