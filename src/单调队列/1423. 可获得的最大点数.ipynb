```python
from typing import List
from math import inf
from functools import cache
```

```python
class Solution:
    def maxScore(self, cardPoints: List[int], k: int) -> int:
        n = len(cardPoints)
        m = n - k
        ans = inf
        res = 0
        s = sum(cardPoints)
        if m == 0:
            return s
        for i, cardPoint in enumerate(cardPoints):
            res += cardPoint
            left = i - m + 1
            if left < 0:
                continue
            ans = min(ans, res)
            res -= cardPoints[left]
        return s - ans
```

```python
class Solution:
    # 内存超出限制
    def maxScore(self, cardPoints: List[int], k: int) -> int:
        @cache
        def dfs(i: int, j: int, c: int) -> int:
            if c == 1:
                return max(cardPoints[i], cardPoints[j])
            return max(dfs(i + 1, j, c - 1) + cardPoints[i], dfs(i, j - 1, c - 1) + cardPoints[j])
        return dfs(0, len(cardPoints) - 1, k)
```
