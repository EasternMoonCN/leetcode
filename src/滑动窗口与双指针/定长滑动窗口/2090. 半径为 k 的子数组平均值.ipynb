```python
from typing import List
```

```python
class Solution:
    def getAverages(self, nums: List[int], k: int) -> List[int]:
        n = len(nums)
        res = [-1] * n
        total = 0

        for i, num in enumerate(nums):
            total += num
            # 长度不足2 * k + 1
            if i < 2 * k:
                continue
            # 具有长度2 * k + 1后，计算i - k 的位置
            res[i - k] = total // (2 * k + 1)
            total -= nums[i - 2 * k]
        return res
```
