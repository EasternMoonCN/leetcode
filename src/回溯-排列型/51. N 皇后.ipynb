```python
class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        ans = []
        path = []
        cols = []
        fDiag = []
        sDiag = []
        def valid(i, j):
            if j in cols or i + j in fDiag or j - i in sDiag:
                return False
            return True

        def generate(j):
            s = ['.'] * n
            s[j] = 'Q'
            return "".join(s)

        def dfs(i):
            if i == n:
                ans.append(path.copy())
                return
            for j in range(n):
                ## valid?
                if not valid(i, j):
                    continue
                cols.append(j)
                fDiag.append(i + j)
                sDiag.append(j - i)
                path.append(generate(j))
                dfs(i + 1)
                path.pop()
                cols.pop()
                fDiag.pop()
                sDiag.pop()
        dfs(0)
        return ans
```

```python
class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        ans = []
        path = []
        cols = [False] * n
        fDiag = [False] * (2*n - 1)
        sDiag = [False] * (2*n - 1)

        def generate(j):
            s = ['.'] * n
            s[j] = 'Q'
            return "".join(s)

        def dfs(i):
            if i == n:
                ans.append(path.copy())
                return
            for j in range(n):
                if cols[j] or fDiag[i + j] or sDiag[j - i]:
                    continue
                cols[j] = fDiag[i + j] = sDiag[j - i] = True
                path.append(generate(j))
                dfs(i + 1)
                path.pop()
                cols[j] = fDiag[i + j] = sDiag[j - i] = False
        dfs(0)
        return ans
```
