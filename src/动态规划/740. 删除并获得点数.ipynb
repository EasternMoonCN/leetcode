```python
class Solution:
    def deleteAndEarn(self, nums: List[int]) -> int:
        n = max(nums) + 1
        # 构造nums中元素num在数组a下标中的值
        # 转换为a数组中不能选相邻元素
        a = [0] * n
        for num in nums:
            a[num] += num
        # 最后按打家劫舍思路求解
        dp = [0] * (n + 2)
        for i in range(n):
            dp[i + 2] = max(dp[i + 1], a[i] + dp[i])
        return dp[n + 1]
```

```python
class Solution:
    def deleteAndEarn(self, nums: List[int]) -> int:
        cnt = Counter(nums)
        d = dict(cnt)
        a = list(cnt)
        a.sort()
        n = len(a)
        dp = [0] * (n + 1)
        dp[1] = a[0] * d[a[0]]
        for i in range(1, n):
            if a[i] - 1 == a[i - 1]:
                dp[i + 1] = max(dp[i], d[a[i]]*a[i] + dp[i - 1])
            else:
                dp[i + 1] =  d[a[i]]*a[i] + dp[i]
        return dp[n]
```
