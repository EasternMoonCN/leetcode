```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        left = 0
        ans = []
        heap = []
        for right, num in enumerate(nums):
            # -num转成最大堆
            heapq.heappush(heap, (-num, right))
            if right - left + 1 < k:
                continue
            ## 此时满足窗口k
            # 先看一眼最大值在哪
            max_val, max_index = heap[0]
            ans.append(-max_val)
            heap.remove((-nums[left], left))
            ## 重建堆
            heapq.heapify(heap)
            left += 1
        return ans
```

```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        queue = []  # 单调递减
        ans = []
        for right, num in enumerate(nums):
            # 1. 入队
            while len(queue) != 0 and nums[queue[-1]] < num:
                queue.pop()
            ## 入队的元素，必须符合递减，所以上面先要出队
            queue.append(right)
            # 2. 出队
            if right - queue[0] + 1 > k:
                queue.pop(0)
            # 3. 结果
            if right >= k - 1:  # 从第k - 1个元素开始记录
            # if right - queue[0] + 1 == k: # 这种不对，因为堆顶不是窗口
                ans.append(nums[queue[0]])
        return ans
```
