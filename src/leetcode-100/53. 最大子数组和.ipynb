```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        self.res = float('-inf')
        @cache
        def dfs(i: int) -> int:
            if i == 0:
                self.res = max(self.res, nums[0])
                return nums[0]
            r = max(0, dfs(i - 1)) + nums[i]
            self.res = max(self.res, r)
            return r
        dfs(len(nums) - 1)
        return self.res
```

```python
class Solution:
    # dp[i] 以i为结束元素的最大子数组和
    def maxSubArray(self, nums: List[int]) -> int:
        n = len(nums)
        dp = [0] * (n + 1)
        for i, num in enumerate(nums):
            dp[i + 1] = max(dp[i], 0) + num
        return max(dp)
```

```python
class Solution:
    # dp[i] 以i为结束元素的最大子数组和
    def maxSubArray(self, nums: List[int]) -> int:
        n = len(nums)
        max_pre, ans = 0, -inf
        for i, num in enumerate(nums):
            max_pre = max(max_pre, 0) + num
            ans = max(max_pre, ans)
        return ans
```
