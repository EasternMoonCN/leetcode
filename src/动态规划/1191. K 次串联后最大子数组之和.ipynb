```python
class Solution:
    def kConcatenationMaxSum(self, arr: List[int], k: int) -> int:
        n = len(arr)
        dp = [0] * (n * k + 1)
        res = 0 # 可以一个不选
        MAX_VAL = 10**9 + 7
        for i in range(n * k):
            dp[i + 1] = max(arr[i % n], dp[i] + arr[i % n])
            res = max(dp[i + 1], res)
        return res % MAX_VAL
```

```python
class Solution:

    def maxSubArray(self, nums: List[int], k: int) -> int:
        n = len(nums)
        res = 0
        # 前缀和
        pre = 0
        # 最小前缀和
        min_pre = 0
        for _ in range(k):
            for x in nums:
                pre += x
                res = max(res, pre - min_pre)
                # 先不选
                min_pre = min(min_pre, pre)
        return res

    def kConcatenationMaxSum(self, arr: List[int], k: int) -> int:
        if k == 1:
            return self.maxSubArray(arr, 1)
        ans = self.maxSubArray(arr, 2)
        ans += max(sum(arr), 0) * (k - 2)
        return ans % 1_000_000_007
```
