```python
class Solution:
    # O(2^N)
    def subsets(self, nums: List[int]) -> List[List[int]]:
        n = len(nums)
        ans = []
        path = []
        def dfs(i: int):
            if i == n:
                ans.append(path.copy())
                return
            # 选
            path.append(nums[i])
            dfs(i + 1)
            path.pop()
            # 不选
            dfs(i + 1)
        dfs(0)
        return ans
```

```python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        n = len(nums)
        ans = []
        path = []
        def dfs(i: int):
            # 没有长度要求，每次便利到的都是答案
            ans.append(path.copy())
            for j in range(i, n):
                path.append(nums[j])
                dfs(j + 1)
                path.pop()
        dfs(0)
        return ans
```
