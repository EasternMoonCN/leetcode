```python
class Solution:
    ## 长度变短时，遍历更小的边界，尝试找下一个更大的边界
    def maxArea(self, height: List[int]) -> int:
        l = 0
        r = len(height) - 1
        ans = 0
        while l < r:
            ans = max(ans, min(height[l], height[r]) * (r - l))
            # height[l] 与右边的任意垂线都无法组成一个比 ans 更大的面积，因为l位置最小了
            if height[l] < height[r]:
                l += 1
            # height[r] 与左边的任意垂线都无法组成一个比 ans 更大的面积
            else:
                r -= 1
        return ans
```
