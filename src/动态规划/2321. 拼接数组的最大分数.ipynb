```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        n = len(nums)
        dp = [0] * (n + 1)
        res = 0
        for i in range(n):
            dp[i + 1] = max(dp[i] + nums[i], nums[i])
            res = max(dp[i + 1], res)
        return res

    # res = s - n1[l:r] + n2[l:r] = s + (n2[l] - n1[l]) + ... + (n2[r] - n1[r]) = s + d[l] + .. + d[r]
    def maximumsSplicedArray(self, nums1: List[int], nums2: List[int]) -> int:
        sum1 = sum(nums1)
        sum2 = sum(nums2)
        diff1 = [n2 - n1 for n1, n2 in zip(nums1, nums2)]
        diff2 = [n1 - n2 for n1, n2 in zip(nums1, nums2)]
        return max(sum1 + self.maxSubArray(diff1), sum2 + self.maxSubArray(diff2))
```
