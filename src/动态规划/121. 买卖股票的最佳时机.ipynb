```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        n = len(prices)
        # dp[i][j]:第i天买入，j天卖出，j > i的利润
        # dp = [[0] * n for _ in range(n)]
        res = 0
        for i in range(n):
            for j in range(i + 1, n):
                res = max(res, prices[j] - prices[i])
        return res
```
