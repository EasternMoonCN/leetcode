```python
class Solution:
    # O(kCkn)
    def combine(self, n: int, k: int) -> List[List[int]]:
        ans = []
        path = []
        #当前需要从[1, i]选一个
        def dfs(i):
            if len(path) == k:
                ans.append(path.copy())
                return
            for j in range(1, i + 1):
                if i + len(path) < k: # 个数无法满足
                    return
                path.append(j)
                dfs(j - 1)
                path.pop()
        dfs(n)
        return ans
```
